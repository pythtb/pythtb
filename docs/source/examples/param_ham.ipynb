```python
from pythtb import TBModel, Lattice
import numpy as np
```

```python
lat = Lattice(lat_vecs=[[1,0],[0,1]], orb_vecs=[[0,0],[1/3, 2/3]], periodic_dirs=[0,1])
tb = TBModel(lattice=lat, spinful=False)

tb.set_onsite("m", ind_i=0)
tb.set_hop("t1", 0, 1, [1, 0])
tb.set_hop(1, 0, 0, [0, 1])
```

```python
x = np.array([0])
y = np.array([1])
np.concatenate((x, y))
```

```python
kpts = tb.k_uniform_mesh([20, 20])
ham = tb.hamiltonian(k_pts=kpts, params={"m": [0, 1, 2], "t1": [-2.5, -2.0, -1.5]})
```

Square lattice with two spinful orbitals

```python
lat = Lattice(
    lat_vecs=[[1, 0], [0, 1]],
    orb_vecs=[[0, 0], [0.5, 0.5]],
    periodic_dirs=[0, 1],
)

tb = TBModel(lattice=lat, spinful=True)

# Symbolic onsite/hopping specifications
tb.set_onsite("mA", ind_i=0)              # orbital A
tb.set_onsite("mB", ind_i=1)              # orbital B
tb.set_hop("t1", 0, 1, [0, 0])            # intra-cell
tb.set_hop("t1", 0, 1, [1, 0])            # +a_x
tb.set_hop("t1", 0, 1, [0, 1])            # +a_y
```

```python
# Evaluate Hamiltonian on a parameter sweep
ham_grid = tb.hamiltonian(
    k_pts=kpts,
    params={
        "mA": np.linspace(-0.2, 0.2, 5),   # first parameter axis (length 5)
        "mB": 0.3,                         # scalar assignment
        "t1": [[0, 1, 1, 0], [0, 0, 1, 0]],          # second axis (length 2)
    },
    flatten_spin_axis=False,
)
```
